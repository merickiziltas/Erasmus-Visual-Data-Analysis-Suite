## **Exercise 3 (Laplacian Eigenmaps vs. t-SNE)**

In this exercise we will focus on Laplacian Eigenmaps and t-SNE, methods that keep negihbors together, instead of trying to preserve global distances between widely separated points.

### **a) Collapsing into a single point**

*How is that prevented in Laplacian Eigenmaps?*

This method aims to minimize $$\sum_{i=1}^n \sum_{j=1}^n w_{ij} \mid \mid y_i-y_j\mid \mid^2$$ where $w_{ij}$ contains weight of edge between $i$ and $j$ and with a normalization that prevents a collapse of all $y_i$ to a point. The normalization is made so that $Y^TY=I$ and so that $Y$ contains $k$ eigenvectors of $L$, the graph Laplacian, for smallest eigenvalues except 0. This normalzation forces fix scale of embedding a non-trivial variance because the solution cannot shrink to zero.

*How is it prevented in t-SNE?*

t-SNE uses probabilities instead of distances: it minimizes $$KL(P\mid \mid Q)=\sum_{i=1} \sum_{j=1} p_{j\mid i} \log{\frac{p_{j\mid i}}{q_{j\mid i}}}$$ where $P$ respresents the fixed similarities in the high-dimensional space and $Q$ the learned similarities in low-dimensional embedding. This works as a cost function because the KL divergence penalizes uniform embeddings. If all points map to the same location, all pairwise distances in the embedding tend to 0, therefore $Q$ has approximately a uniform distribution. But $P$ is not uniform, ceating a large mismatch so the cost is very high.

### **b) Crowded clusters**

t-SNE is more effective at preventing crowding within clusters. As stated in the slides, it abstracts away density and distance information and focuses on preserving local neighborhoods, often revealing cluster structure more clearly. This leads to an expansion of dense regions, making points within clusters more distinguishable. In contrast, Laplacian Eigenmaps preserves local structure without altering densities, so dense regions remain crowded.

### **c) Edge weightning**

*Problems in Laplacian Eigenmaps*

Although k-NN graphs ensure connectivity in Laplacian Eigenmaps, edge weights still depend on distances: $$w_{ij}​=\exp(−\frac{∥x_i​−x_j​∥^2}{\sigma ^2}​).$$ 
Therfore, when neighbors are far apart, distances are larger and weights are closer to 0. So, in regions of varying density, this leads to strong connections in dense areas and weak connections in sparse areas, causing distortions in the embedding.

*How t-SNE avoids it*

t-SNE avoids this problem by adapting the neighborhood scale for each point using perplexity (= effective number of neighbors), effectively normalizing for local density and ensuring a more balanced treatment of neighborhoods.

### **d) Determinism**

In this case, Laplacian Eigenmaps are preferable with respect to determinism, because this method is based on computing the eigenvalues of a ceratin system. Therfore, it produces stable and reproducible embeddings, eventhough they might experience sign flips with trivial transformations. In contrast, t-SNE relies on random initialization and non-convex optimization, which can lead to different results across runs, making it less deterministic.

### **e) Axis**

As we said before, the Laplacian Eigenmaps are based on a eigenvalue problem. Consequently, the embedding axes correspond to eigenvectors of the graph Laplacian and are therefore naturally ordered by their associated eigenvalues. 

In contrast, t-SNE does not define any ordering of axes, because its objective depends only on pairwise similarities.
