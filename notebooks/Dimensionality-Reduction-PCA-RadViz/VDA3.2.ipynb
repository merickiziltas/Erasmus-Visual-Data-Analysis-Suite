**a)** *As we mentioned in the lecture, the star coordinates, and RadViz are two popular multidimensional
data visualization techniques. Why they can also be interpreted as dimensionality reduction
techniques? (1P)*

They can be interpreted as dimensionality reduction techniques because they transform high-dimensional objects into points in the 2d-plane, while ensuring that the output can be used to retrieve in some sense the different original attributes.

**b)** *In general, the authors of this paper prefer star coordinates over RadViz. Briefly explain two
reasons for their preference. (2P)*

One reason the authors prefer SC over RadViz is the fact that RadViz apllies a non-linear transformation to the original data, which in term implies that, in contrast to SC, it cannot be used directly to recreate popular linear data reduction techniques, such as LDA or PCA, which are in some sense specific cases of SC.

Another reason the authors provide is the uneven distortion of the data depending of its location in the data space, which implies that data clusters with similar shape can have vastly different appearences in the resulting plot.

**c)** *Despite this, the authors mention a specific use case where they consider RadViz to be superior
to star coordinates. Briefly explain what this use case is and why they prefer RadViz in this case.
(2P)*


The authors mention that RadViz has an advantage in the detection of sparse data, which they define as data points which have a really high value in one particular attribute. The sample is then shown near the anchor point for that attribute in the RadViz plot, while the linear interference of the other dimensions when using SC makes the data more evenly spread throughout. 

**d)** *Neither RadViz nor star coordinates provide a one-to-one mapping between a data point’s highdimensional location and its two-dimensional projection. Interactively modifying the anchor points
or axes, respectively, can reduce the resulting ambiguities. However, the paper mentions two
examples in which, unlike star coordinates, RadViz introduces ambiguities that cannot be resolved
with any re-arrangement of the anchor points. Point out these two examples. Could the extended
RadViz that we learned about in the lecture resolve them? (3P)*


The authors mention two cases of datasets in which the RadViz method causes ambiguities that cannot be resolved by moving the anchor points.

The first of those is due to the normalization of the data and the non-linear effect of the transformation. This can cause two groups on which there already existed some overlap to come even closer together in the transformed data and thus difficult to discern (they use the example of *Iris* dataset). The extended RadViz method could alleviate this problem, because the influence of the different anchor points can be directly observed, which would help in noticing the difference in the separating variable for these groups.

The second example they provide happens when all the variables are positively correlated. Then, as some variables have to point in opposite directions, an increase in both at the same time cannot be observed in the plot (they give the example of the *US breakfast cereal* dataset). The use of the extended RadViz method could help in this case, because the influence of several opposing variables can be noticed by a proportional increase in the length of the *springs* pointing to the dataset. On the other hand, the fact that points with such different values can end really close in the plot would make it hard to differentiate the individual points which are being pointed.

**e)** *Several methods have been proposed to optimize the locations of anchor points in RadViz to
obtain an improved separation of classes in the resulting projection. How did two such algorithms
compare to supervised linear dimensionality reduction techniques in the experiments reported in
this paper, in terms of (i) computational effort and (ii) achieved class separation? (2P)*


The authors mention the use of two different algorithms that try to reduce the classification error rate for a RadViz plot, namely the *CDL algorithm*, and another method described in a paper of *Albuquerque et al.*, and compare their performance to the well-known LDA method.

In terms of computational complexity, both RadViz methods have an superexponential running time in terms of the number of dimensions, while the LDA method runs in linear time. This meant that even for a relatively low level of dimensions (in this case d = 9), the execution time was considerably longer for the former (they mention 22.4 hours of runnung time), as for the latter method (in a few seconds).

In terms of achieved class separation, both mentioned methods fare relatively poorly (13.46% and 4.49% error rates respectively) with respect to the LDA method (0.56%), according to the 5-nearest neighbor classifier test.

**f)** *Briefly explain why anchor points in RadViz are commonly chosen such that they form a corner
of the overall convex hull. Is this property also met for the star coordinate axis configuration in
Fig. 12 (a)? Is it true in Fig. 12 (b)? (2P)*


Anchor points are chosen to form corners of the convex hull of the plot because otherwise the legibility of the plot would be heavily compromised, as attraction to several exterior anchor points could be mistaken for attraction to an interior anchor point, or cause that sparse points appear in the same region as the central cluster of a distribution. As every point is plotted in this convex hull, having a thin or otherwise odd-shaped polygon would make the plot hard to analyze.

This property is not always verified in the case of SC without causing a decrease in the plot's readability. In figure 12 a) we see that there are several attributes clumped around the origin, clearly immersed by the hull of the longer vectors around them. Even in the case of figure 12 b) this condition isn't verified as the vector representing *Alcohol* is contained in the triangle described by *Ash*, *Proline* and *Color Intensity*.
